```python
# Create a working copy
df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Remove extra spaces from text columns
for col in df_clean.select_dtypes(include='object').columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Check missing values again
print(df_clean.isnull().sum())

# Check final dimensions
print("Cleaned Dataset Shape:", df_clean.shape)
```

```python
# Check missing values in each column

missing_values = df_clean.isnull().sum()

print(missing_values)
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean)
) * 100

print(missing_percentage.round(2))
```

```python
# Identify categorical and numerical columns

categorical_columns = df_clean.select_dtypes(
    include=['object']
).columns

numerical_columns = df_clean.select_dtypes(
    include=['number']
).columns

# Fill missing categorical values

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

# Fill missing numerical values with median

for col in numerical_columns:
    if df_clean[col].notna().any():
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

print("Missing values handled successfully!")
```

```python
print(df_clean.dtypes)
```

```python
# Convert timestamp column

if 'Timestamp' in df_clean.columns:
    df_clean['Timestamp'] = pd.to_datetime(
        df_clean['Timestamp'],
        errors='coerce'
    )

print("Timestamp conversion completed.")
```

```python
if 'Timestamp' in df_clean.columns:
    df_clean['Year'] = df_clean['Timestamp'].dt.year
    df_clean['Month'] = df_clean['Timestamp'].dt.month
    df_clean['Day'] = df_clean['Timestamp'].dt.day

df_clean.head()
```

```python
df_clean['Timestamp'] = pd.to_datetime(
    df_clean['Timestamp'],
    errors='coerce'
)

monthly_posts = (
    df_clean.dropna(subset=['Timestamp'])
    .set_index('Timestamp')
    .resample('MS')
    .size()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_posts.index,
    monthly_posts.values,
    marker='o'
)

plt.title('Monthly Social Media Posting Trends')
plt.xlabel('Month')
plt.ylabel('Number of Posts')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import re
from collections import Counter

hashtags = []

for value in df_clean['Hashtags'].dropna():
    hashtags.extend(
        re.findall(r'#([A-Za-z0-9_]+)', str(value))
    )

hashtag_counts = Counter(hashtags).most_common(10)

if hashtag_counts:
    hashtag_df = pd.DataFrame(
        hashtag_counts,
        columns=['Hashtag', 'Count']
    )

    plt.figure(figsize=(10, 5))

    sns.barplot(
        data=hashtag_df,
        x='Count',
        y='Hashtag'
    )

    plt.title('Top 10 Most Frequently Used Hashtags')
    plt.xlabel('Frequency')
    plt.ylabel('Hashtag')
    plt.tight_layout()
    plt.show()
else:
    print("No hashtags with # were found. Check the hashtag format.")
```
